Part 1: Conceptual Check

Question 1: What is the key difference between a Pandas Series and a DataFrame?
Ans: A series is 1-dimensional whereas a DataFrame is 2-dimensional, similar to a table.

Question 2: You run df.loc[0:3]. How many rows do you get? What about df.iloc[0:3]?
Ans: df.loc returns 4 rows from 0 to 4. df.iloc returns 3 rows from 0 to 2.

Question 3: You have a DataFrame with a price column stored as object type (strings like "£45.99"). What problem does this cause, and how do you fix it?
Ans: It will be difficult to perform calculations on an object. Instead, I would divide the string into 2 parts: 1 for currency stored as a string, and 1 for the price stored as a float.

Question 4: When would you use .apply() instead of a direct vectorised operation like df['col'] * 2?
Ans: Vectorised operations are faster and should be used where possible. .apply() is used when row-level logic involves multiple columns or when no vectorised equivalent exists.

Question 5: How do you select all numeric columns from a DataFrame, excluding float types?
Ans: df.select_dtypes(include=['number'], exclude=['float'])

```python
# Part 2: Practical Challenge

import pandas as pd

books = pd.DataFrame({
    'title': ['The Pragmatic Programmer', 'Clean Code', 'Python Crash Course',
              'Data Science from Scratch', 'Designing Data-Intensive Apps'],
    'author': ['Hunt & Thomas', 'Robert Martin', 'Eric Matthes',
               'Joel Grus', 'Martin Kleppmann'],
    'genre': ['Tech', 'Tech', 'Tech', 'Data', 'Data'],
    'price': [45.99, 38.50, 29.99, 42.00, 55.00],
    'units_sold': [120, 95, 200, 80, 60],
    'in_stock': [True, True, False, True, False]
})


print(books.info(), "\n\n")
# The number of non-null values in each column is 5

print(books.describe(), "\n\n")
books['revenue'] = books['price'] * books['units_sold']
print(books)
print("The total revenue from all books is: £", f"{books['revenue'].sum():.2f}")
```

```python
print(books.loc[:, ['title', 'revenue']])
print(books.iloc[-3:, :3])
print(books[(books['in_stock'] == True) & (books['price'] > 40)])

top_book = books.loc[books['revenue'].idxmax()]
print("The book with the highest revenue is:", top_book['title'])
```

```python
# Part 2: Challenge 3: Transform, Sort and Rank

def classify_price(price):
    if price < 35:
        return 'Budget'
    elif price <= 50:
        return 'Standard'
    else:
        return 'Premium'

books['price_tier'] = books['price'].apply(classify_price)
print(books[['title', 'price', 'price_tier']])

books.sort_values(by='revenue', ascending=False, inplace=True)
print(books)

books['sales_rank'] = books['units_sold'].rank(ascending=False)
print(books)

def genre_encode(genre):
    if genre == 'Tech':
        return 0
    elif genre == 'Data':
        return 1
    else:
        return -1

books['genre_encoded'] = books['genre'].apply(genre_encode)
print(books[['title', 'genre_encoded']])
```
